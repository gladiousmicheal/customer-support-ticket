```python
storage_code = '''
import json
from pathlib import Path

DATA_FILE = Path(__file__).parent / "tickets.json"


def load_tickets():
    """Load saved tickets from the JSON file."""
    if not DATA_FILE.exists() or DATA_FILE.stat().st_size == 0:
        return []

    try:
        with DATA_FILE.open("r", encoding="utf-8") as file:
            return json.load(file)
    except json.JSONDecodeError:
        return []


def save_tickets(tickets):
    """Save tickets to the JSON file."""
    with DATA_FILE.open("w", encoding="utf-8") as file:
        json.dump(tickets, file, indent=4)
'''

(PROJECT_DIR / "storage.py").write_text(
    storage_code.lstrip(), encoding="utf-8"
)

print("storage.py saved successfully!")
```

```python

from pathlib import Path

PROJECT_DIR = Path("/Users/gladious/Documents/gCloudPractice/customer-support-ticket/Untitled Folder")
test_file = PROJECT_DIR / "test_app.py"

new_test = '''

def test_reject_invalid_priority(monkeypatch, tmp_path, capsys):
    ticket_file = tmp_path / "tickets.json"
    ticket_file.write_text("[]", encoding="utf-8")

    monkeypatch.setattr(
        app,
        "load_tickets",
        lambda: json.loads(ticket_file.read_text(encoding="utf-8"))
    )

    def save(tickets):
        ticket_file.write_text(json.dumps(tickets), encoding="utf-8")

    monkeypatch.setattr(app, "save_tickets", save)

    answers = iter(["Order delayed", "Not received", "Urgent"])
    monkeypatch.setattr("builtins.input", lambda _: next(answers))

    app.create_ticket()

    tickets = json.loads(ticket_file.read_text(encoding="utf-8"))

    assert tickets == []
    assert "Invalid priority" in capsys.readouterr().out
'''

with test_file.open("a", encoding="utf-8") as file:
    file.write(new_test)

print("Second test added successfully!")
```
